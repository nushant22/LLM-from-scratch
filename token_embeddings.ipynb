Let's illustrate how the token ID to embedding vector conversion works with a hands-on
example. Suppose we have the following four input tokens with IDs 2, 3, 5, and 1:

```python
import torch

input_ids = torch.tensor([2, 3, 5, 1])
```

For the sake of simplicity and illustration purposes, suppose small vocabulary of
only 6 words (instead of the 50,257 words in the BPE tokenizer vocabulary), and I want
to create embeddings of size 3 (in GPT-3, the embedding size is 12,288 dimensions):

Using the vocab_size and output_dim, I can instantiate an embedding layer in PyTorch,
setting the random seed to 123 for reproducibility purposes:

```python
vocab_size = 6
output_dim = 3

torch.manual_seed(123)
embedding_layer = torch.nn.Embedding(vocab_size, output_dim)
```

The print statement in the code prints the embedding layer's underlying
weight matrix:

```python
print(embedding_layer.weight)
```

The weight matrix of the embedding layer contains small, random values.
These values are optimized during LLM training as part of the LLM optimization itself. Moreover, we can see that the weight matrix has six rows
and three columns. There is one row for each of the six possible tokens in the vocabulary.
And there is one column for each of the three embedding dimensions.

After instantiating the embedding layer, let's now apply it to a token ID to obtain the
embedding vector:

```python
print(embedding_layer(torch.tensor([3])))
```

Comparing the embedding vector for token ID 3 to the previous embedding matrix, we
see that it is identical to the 4th row (Python starts with a zero index, so it's the row
corresponding to index 3). In other words, the embedding layer is essentially a look-up
operation that retrieves rows from the embedding layer's weight matrix via a token ID.

Previously, we have seen how to convert a single token ID into a three-dimensional
embedding vector. Let's now apply that to all four input IDs we defined earlier
(torch.tensor([2, 3, 5, 1])):

```python
print(embedding_layer(input_ids))
```

Each row in this output matrix is obtained via a lookup operation from the embedding
weight matrix
